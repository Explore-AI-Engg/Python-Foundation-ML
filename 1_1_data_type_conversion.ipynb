```python
x = 10
print(x)
print(type(x))

x = float(x) # ---> convert to float 
print(x)
print(type(x))
```

```python
y = 5.6
print(y)
print(type(y))

y = int(y) # ---> convert to int
print(y)
print(type(y))
```

```python
z = 10
print(z)
print(type(z))

z = str(z) # ---> convert to string   
print(z)
print(type(z))
```

```python
age = input("Enter your age: ")
print(f"You entered: {age}")
print(type(age))
```
