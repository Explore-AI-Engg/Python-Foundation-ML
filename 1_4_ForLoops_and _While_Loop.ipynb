# 2.while loop 

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

#3. Loop Contains - break:    

```python
for i in range(0, 15):
    if  i == 4:
        break
    print(i)
```

```python
# Continue - skip an itration
for i in range(0, 10):
    if i % 2 == 0:
        continue
    print(i)
```
