# List - Ordered & Changeable (Mutable) 

```python
# Creating a list 

fruits = ["apple", "banana", "cherry"]
print(fruits)
```

```python
# Accessing list elements using index
print(fruits[0])  # Output: apple
print(fruits[1])  # Output: banana
print(fruits[2])  # Output: cherry
```

```python
# change an elements values
fruits[1] = "blueberry"
print(fruits)
```

```python
a = [1, 2, 3, False, "Hellow"]
print(a)
```

```python
print(type(a))
```

```python
a.append("World")
print(a)
```

```python
# remove an element
a.remove("Hellow")
```

```python
# find the hunber of the elements in the list
print(len(a))
```

# tupel - ordered unchanged (imutable)

```python
# tuple of colors
colors = ("red", "green", "blue")
print(colors)   
print(type(colors))
```

```python
print(colors[0])  # Output: red
print(colors[1])  # Output: green
print(colors[2])  # Output: blue
```

```python
# Rtying to change an element in a tuple will result in an error
# colors[1] = "yellow"  # This will raise a TypeError 
```

```python
for i in colors:
    print(i)
```

```python
a = [1, 2, 3]
b = tuple(a)

print(b)
print(type(b))

print(a)
print(b)

```
