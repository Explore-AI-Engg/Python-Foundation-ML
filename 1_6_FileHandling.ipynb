# opening a file and read a File

```python
#open a file in read node
file = open("./SupportedFiles/notes.txt", "r")
#read the content of the file
content = file.read()
print(content)
#close the file
file.close()
```

# 2. Using with to read (auto-closing)

```python
with open("./SupportedFiles/notes.txt", "r") as file:
    content = file.read()
    print(content)
```

# 3. Writing to a File

```python
with open("./SupportedFiles/notes2.txt", "w") as file:
    file.write("This is a new line added to the file.\n")
    file.write("This is another line added to the file.\n") 
```

```python
with open("./SupportedFiles/notes2.txt", "a") as file:
    file.write("This is a new line added to Existing content.\n")
    file.write("This is a new line added to the file.\n")
    file.write("This is another line added to the file.\n") 
```

```python
with open("./SupportedFiles/notes2.txt", "r") as file:
  for line in file:
    print(line.strip())
```
