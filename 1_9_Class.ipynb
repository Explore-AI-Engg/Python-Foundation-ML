### 4. USe Class Attributes

Attributes are varibales the belong to a class, Let's add and use attributes in our Dog Class

```python
# add attribute to the Dog class
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        print("Woof!")  
```

```python
my_dog = Dog("Buddy", age=3)
print(my_dog.name)
print(my_dog.age)
```

```python
my_dog1 = Dog("Tomy", age=12)
print(my_dog1.name)
print(my_dog1.age)
```

# 5. DemonStrate Inheritance

Inheritance allows once class to use the properities and method of anothe class. Let's create a subclass that inherits from Dog.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        print("Woof!")
```

```python
# creating a subclass that inherits from Dog
class Puppy(Dog):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed
    def play(self):
        print("The puppy is playing.")
```

```python
my_Puppy = Puppy("Charlie", age=1, breed="Golden Retriever")
print(my_Puppy.name)
print(my_Puppy.age)
print(my_Puppy.breed)
```
